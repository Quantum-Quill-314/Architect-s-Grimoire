```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"C:\Users\HP\Downloads\Processed DataFrame.csv", sep=';')
```

```python
df.columns
```

```python
sns.set_theme(style = "darkgrid")
plt.style.use('dark_background')
```

```python
#histogram
plt.figure(figsize=(10, 6))
sns.histplot(df['pChEMBL Value'], kde=True, color='teal')
plt.title('Distribution of pChEMBL Value')
plt.xlabel('pChEMBL Value')
plt.ylabel('Frequency')
plt.savefig('histogram_pchembl.png')
```

```python
#boxplot
plt.figure(figsize=(10, 6))
sns.boxplot(x='Activity_Class', y='Molecular Weight', data=df, palette='viridis')
plt.title('Molecular Weight Distribution by Activity Class')
plt.xlabel('Activity Class (0: Inactive, 1: Active)')
plt.ylabel('Molecular Weight')
plt.savefig('boxplot_mw_activity.png')
```

```python
#Heatmap correlation matrix
scaled_cols = ['std_Molecular Weight', 'std_AlogP', 'std_Ligand Efficiency BEI']
corr_matrix = df[scaled_cols].corr()

plt.figure(figsize=(10, 6))
sns.heatmap(corr_matrix, annot=True, cmap='mako', fmt=".2f", square=True)
plt.title('Correlation Matrix of Scaled Features')
plt.savefig('heatmap_correlation.png')
```

```python
#Contour Plot
plt.figure(figsize=(10, 8))
sns.kdeplot(data=df, x='Molecular Weight', y='AlogP', fill=True, cmap='twilight', thresh=0.05)
plt.title('2D KDE: Molecular Weight vs. AlogP (Topological Goldilocks Zone)')
plt.xlabel('Molecular Weight')
plt.ylabel('AlogP')
plt.savefig('kde_mw_alogp.png')
```

```python
#scatter plot
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='Molecular Weight', y='Ligand Efficiency BEI',
                hue='Activity_Class', palette='Set2', alpha=0.6)
plt.title('Ligand Efficiency BEI vs. Molecular Weight')
plt.xlabel('Molecular Weight')
plt.ylabel('Ligand Efficiency BEI')
plt.legend(title='Activity Class')
plt.savefig('scatter_mw_lei.png')
```

```python
#violin plot
plt.figure(figsize=(10, 6))
sns.violinplot(x='Activity_Class', y='AlogP', data=df, palette='Dark2', split=True)
plt.title('AlogP Density Distribution by Activity Class')
plt.xlabel('Activity Class')
plt.ylabel('AlogP')
plt.savefig('violin_alogp_activity.png')

plt.show()
```
